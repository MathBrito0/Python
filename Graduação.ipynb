```python
from Crypto.Cipher import AES
from base64 import b64decode
import hashlib
import json

def testar_chave(pack_json, senha):
  pack = json.loads(pack_json)

  nonce = b64decode(pack["nonce"])
  tag = b64decode(pack["tag"])
  ct = b64decode(pack["ct"])

  key = hashlib.sha256(senha.encode()).digest()
  aes = AES.new(key, AES.MODE_GCM, nonce=nonce)

  try:
    pt = aes.decrypt_and_verify(ct, tag)
    # Se chegou aqui, a senha está correta.
    # O GCM só retorna o texto se:
    # 1. a chave estiver certa
    # 2. o nonce/tag/ciphertext baterem
    return pt.decode()

  except Exception:
    # Se caiu aqui, a tentativa falhou.
    return None

def main():
  # Texto criptografado
  pack_json = '{"nonce": "vLunt5B4eMtDIL7psEF3+A==", "tag":"fVrbvAG85PXfCu4Os8stKA==", "ct": "ipY68YAWkPHaXdy2liqv6gkDSQ=="}'

  contador = 0
  var = 'abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789'
  for i in range(62):
    for j in range(62):
      for k in range(62):
        for l in range(62):
          variavel = f'{var[i]}{var[j]}{var[k]}{var[l]}'
          texto = testar_chave(pack_json, variavel)
          contador +=1
          if texto is not None:
            print(f'a senha é {variavel}')
            print(f'o segredo é {texto}')
            return

main()
```

### Alterações Realizadas no Script de Força Bruta

| Parte do Código | Antes | Depois | Motivo |
| :--- | :--- | :--- | :--- |
| **Log de tentativas** | `print(variavel)` | `# print(variavel)` | Evita que o navegador trave e que o log real (a senha) seja perdido no meio de milhões de linhas. |
| **Interrupção** | Continuava o loop | `return` | Interrompe a execução assim que a senha correta é encontrada, economizando tempo de processamento. |
